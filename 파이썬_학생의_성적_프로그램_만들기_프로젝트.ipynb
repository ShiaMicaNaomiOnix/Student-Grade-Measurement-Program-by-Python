```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

grade = pd.read_csv('student_scores_clear_utf8.csv', encoding='utf-8-sig')

grade.dropna(inplace=True)
student_number = np.arange(len(grade))

kor = grade["국어"]
eng = grade["영어"]
math = grade["수학"]
avg = (kor + eng + math) / 3

# 4개의 서브플롯을 만들기
fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True, sharey=True)

# 첫번째 그래프 - 국어 점수
axes[0, 0].hist(student_number, kor, color='blue', alpha=0.7, label='국어')
axes[0, 0].set_title("전체 학생의 국어 점수")
axes[0, 0].set_ylabel("점수")
axes[0, 0].legend()

# 두번째 그래프 - 영어 점수
axes[0, 1].hist(student_number, eng, color='orange', alpha=0.7, label='영어')
axes[0, 1].set_title("전체 학생의 영어 점수")
axes[0, 1].legend()

# 세번째 그래프 - 수학 점수
axes[1, 0].hist(student_number, math, color='green', alpha=0.7, label='수학')
axes[1, 0].set_title("전체 학생의 수학 점수")
axes[1, 0].set_xlabel("학생 번호")
axes[1, 0].set_ylabel("점수")
axes[1, 0].legend()

# 네번째 그래프 - 평균 점수
axes[1, 1].hist(student_number, avg, color='purple', alpha=0.7, label='평균')
axes[1, 1].set_title("각 학생의 평균 점수")
axes[1, 1].set_xlabel("학생 번호")
axes[1, 1].legend()

# x축 눈금 레이블을 학생 번호로 지정
for ax in axes.flat:
    ax.set_xticks(student_number)

plt.tight_layout()
plt.show()
```

```python
#학생 점수 찾기
from os import name
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# CSV 읽기
grade = pd.read_csv('student_scores_clear_utf8.csv', encoding='utf-8-sig')
grade.dropna(inplace=True)
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

# 평균 점수 컬럼 추가
grade["평균"] = (grade["국어"] + grade["영어"] + grade["수학"]) / 3

# 특정 학생 번호 입력
student_index = int(input("학생 번호를 입력하세요 (0부터 시작): "))


# 학생 정보
subjects = ['국어', '영어', '수학']
scores = grade.loc[student_index, subjects].values
avg_score = grade.loc[student_index, '평균']


# 그래프1
ax[0].bar(subjects, scores, color=['skyblue', 'salmon', 'lightgreen'], edgecolor='black', label=['국어', '영어', '수학'])
ax[0].bar(['평균'], avg_score, color='purple', edgecolor='black', label='평균 점수')
ax[0].set_ylim(0, 105)
ax[0].set_title(f"{grade.loc[student_index, '이름']} 학생의 점수")
ax[0].set_ylabel('점수')
ax[0].legend()

#그래프2
# 두 번째 그래프 - 최고/최저 점수 + 과목명 표시
score_max = np.max(scores)
score_min = np.min(scores)
score_max_subject = subjects[np.argmax(scores)]
score_min_subject = subjects[np.argmin(scores)]
labels = [f"최고({score_max_subject})", f"최저({score_min_subject})"]
ax[1].bar(labels, [score_max, score_min], color=['red', 'blue'], edgecolor='black')
ax[1].set_ylim(0, 105)
ax[1].set_title(f"{grade.loc[student_index, '이름']} 학생의 최고/최저 점수 과목")
ax[1].set_ylabel('점수')
ax[1].legend()

plt.tight_layout()
plt.show()
```

```python
#전체의 평균 국어, 영어, 수학, 평균 점수 보여주기

grade["평균"] = (grade["국어"] + grade["영어"] + grade["수학"]) / 3

# 전체 과목별 평균 점수
mean_kor = grade["국어"].mean()
mean_eng = grade["영어"].mean()
mean_math = grade["수학"].mean()
mean_avg = grade["평균"].mean()

print(f"전체 국어 평균: {mean_kor:.2f}")
print(f"전체 영어 평균: {mean_eng:.2f}")
print(f"전체 수학 평균: {mean_math:.2f}")
print(f"전체 평균 점수: {mean_avg:.2f}")

# 그래프로 한눈에 보기
subjects = ["국어", "영어", "수학", "전체 평균"]
means = [mean_kor, mean_eng, mean_math, mean_avg]

plt.figure(figsize=(8, 6))
bars = plt.bar(subjects, means, color=["blue", "orange", "green", "purple"],  edgecolor='black')

# 각 막대 위에 값 표시
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.5, f"{height:.2f}", ha='center', va='bottom', fontsize=10)

plt.title("전체 과목별 평균 점수")
plt.ylabel("평균 점수")
plt.ylim(0, 100)  # 점수 범위
plt.tight_layout()
plt.show()
```
